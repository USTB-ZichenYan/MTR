```python
import torch

# 创建两个不同数据类型的张量
obj_trajs = torch.randn(2, 3, 5,dtype=torch.float32)  # 浮点类型张量
obj_trajs_mask = torch.randint(0, 2, (2, 3), dtype=torch.bool)  # 布尔类型张量

print("原始 obj_trajs 数据类型:", obj_trajs.dtype)
print("原始 obj_trajs_mask 数据类型:", obj_trajs_mask.dtype, obj_trajs_mask)

# 使用 type_as 方法将 obj_trajs_mask 转换为和 obj_trajs 相同的数据类型
obj_trajs_mask_converted = obj_trajs_mask.type_as(obj_trajs)

print("转换后 obj_trajs_mask 数据类型:", obj_trajs_mask_converted.dtype, print(obj_trajs_mask_converted))

# 进行拼接操作
obj_trajs_in = torch.cat((obj_trajs, obj_trajs_mask_converted.unsqueeze(-1)), dim=-1)
print("拼接后的张量形状:", obj_trajs_in.shape)
```

```python
print("obj_trajs_mask: ", obj_trajs_mask.shape, obj_trajs_mask)
# apply self-attn
obj_valid_mask = (obj_trajs_mask.sum(dim=-1) > 0)  # (num_center_objects, num_objects)
print("obj_valid_mask ", obj_valid_mask.shape, obj_valid_mask)
```
